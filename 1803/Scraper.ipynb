```python
import os
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import time

# Folder to save images
save_folder = r"G:\My Drive\Project Mushroom\dataset\Clavulina coralloides"
os.makedirs(save_folder, exist_ok=True)

# Headers to mimic a real browser
headers = {"User-Agent": "Mozilla/5.0"}

def get_soup(url):
    """Fetch and parse HTML of a webpage."""
    try:
        response = requests.get(url, headers=headers, timeout=10)
        response.raise_for_status()
        return BeautifulSoup(response.text, "html.parser")
    except requests.exceptions.RequestException as e:
        print(f"Error fetching {url}: {e}")
        return None

def download_image(img_url, obs_id):
    """Download and save full-size image with a unique name."""
    if not img_url.startswith("http"):
        img_url = urljoin(base_url, img_url)

    img_name = f"{obs_id}_{os.path.basename(img_url).split('?')[0]}"  # Add observation ID to filename
    img_path = os.path.join(save_folder, img_name)

    if os.path.exists(img_path):
        print(f"Skipping (already downloaded): {img_name}")
        return

    try:
        img_data = requests.get(img_url, headers=headers, timeout=10).content
        with open(img_path, "wb") as f:
            f.write(img_data)
        print(f"Downloaded: {img_name}")
    except Exception as e:
        print(f"Failed to download {img_url}: {e}")

def scrape_observation_page(obs_url):
    """Scrape images from a single observation page and fetch full-size images."""
    current_url = obs_url
    obs_id = obs_url.split("/")[-1].split("?")[0]  # Extract observation ID

    while current_url:
        print(f"\nScraping: {current_url}")
        soup = get_soup(current_url)
        if not soup:
            break

        # Find all images and get the full-size URL
        for img in soup.find_all("img"):
            img_src = img.get("src")
            if img_src and "/images/" in img_src:
                full_img_url = img_src.replace("/images/thumb/", "/images/orig/")  # Convert to full-size URL
                download_image(urljoin(current_url, full_img_url), obs_id)

        # Find the "View More Photos" button and follow the link
        more_photos_link = soup.find("a", string="View More Photos")
        if more_photos_link:
            current_url = urljoin(current_url, more_photos_link["href"])
            time.sleep(2)  # Delay to avoid getting blocked
        else:
            break

# Get multiple links from user input at once
user_input = input("\nEnter Mushroom Observer URLs separated by spaces:\n").strip()
urls = user_input.split()

# Process each URL one by one
for url in urls:
    if url.startswith("https://mushroomobserver.org/"):
        scrape_observation_page(url)
        print(f"\nFinished downloading for {url}\n")
    else:
        print(f"Invalid URL skipped: {url}")

print("\nAll downloads completed successfully!")
```
