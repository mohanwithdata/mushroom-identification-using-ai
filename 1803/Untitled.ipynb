```python
import os
import random
import shutil

# Define the paths
original_dataset_dir = r'D:\Project\dataset'
output_dir  = r'G:\My Drive\Project Mushroom\Data\new_dataset'

os.makedirs(output_dir, exist_ok=True)

# Define the number of images per set
train_count = 100
val_count = 20
test_count = 20
total_required = train_count + val_count + test_count

# Create the output directories
train_dir = os.path.join(output_dir, 'train')
val_dir = os.path.join(output_dir, 'val')
test_dir = os.path.join(output_dir, 'test')
os.makedirs(train_dir, exist_ok=True)
os.makedirs(val_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)

# Iterate over each class folder in the original dataset
for class_name in os.listdir(original_dataset_dir):
    class_dir = os.path.join(original_dataset_dir, class_name)
    if os.path.isdir(class_dir):
        # List all image files in the class directory
        images = [f for f in os.listdir(class_dir) if os.path.isfile(os.path.join(class_dir, f))]
        
        # Check if there are enough images
        if len(images) < total_required:
            print(f"Warning: Class '{class_name}' has less than {total_required} images.")
            continue
        
        # Randomly shuffle and select images
        random.shuffle(images)
        train_images = images[:train_count]
        val_images = images[train_count:train_count + val_count]
        test_images = images[train_count + val_count:total_required]
        
        # Create class directories in the output folders
        train_class_dir = os.path.join(train_dir, class_name)
        val_class_dir = os.path.join(val_dir, class_name)
        test_class_dir = os.path.join(test_dir, class_name)
        os.makedirs(train_class_dir, exist_ok=True)
        os.makedirs(val_class_dir, exist_ok=True)
        os.makedirs(test_class_dir, exist_ok=True)
        
        # Copy images to the respective directories
        for image in train_images:
            shutil.copy(os.path.join(class_dir, image), os.path.join(train_class_dir, image))
        for image in val_images:
            shutil.copy(os.path.join(class_dir, image), os.path.join(val_class_dir, image))
        for image in test_images:
            shutil.copy(os.path.join(class_dir, image), os.path.join(test_class_dir, image))

print("Dataset split into training, validation, and test sets.")
```

```python
import os
from PIL import Image

def resize_images_in_directory(directory, size=(224, 224)):
    """
    Resizes all images in the specified directory to the given size.
    """
    for root, _, files in os.walk(directory):
        for file in files:
            if file.lower().endswith(('png', 'jpg', 'jpeg', 'bmp', 'gif')):
                image_path = os.path.join(root, file)
                try:
                    with Image.open(image_path) as img:
                        img = img.resize(size, Image.LANCZOS)
                        img.save(image_path)
                except Exception as e:
                    print(f"Error processing {image_path}: {e}")

def main(base_dir):
    """
    Resizes images in 'train' and 'val' subdirectories of the base directory.
    """
    for subset in ['train', 'val']:
        subset_dir = os.path.join(base_dir, subset)
        if os.path.isdir(subset_dir):
            print(f"Resizing images in {subset_dir}...")
            resize_images_in_directory(subset_dir)
        else:
            print(f"Directory {subset_dir} does not exist.")

if __name__ == "__main__":
    base_dataset_dir = r'G:\My Drive\Project Mushroom\Data\new_dataset'
    main(base_dataset_dir)
    print("Image resizing complete.")
```
