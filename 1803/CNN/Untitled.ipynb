```python
# Paths
source_folder = r'D:\Project\data_kaggle\train'
destination_folder =  r'D:\Project\data_kaggle\aug\train_augmented'  # Folder to store new augmented images

import os
import cv2
import albumentations as A
import numpy as np
from albumentations import HorizontalFlip, Rotate, ColorJitter, Compose

# Ensure destination folder exists
os.makedirs(destination_folder, exist_ok=True)

# Define augmentation pipeline
augmentation = Compose([
    HorizontalFlip(p=0.5),    # 50% chance of horizontal flip
    Rotate(limit=30, p=0.7),  # Rotate within ±30 degrees
    ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1, p=0.5)  # Random color jitter
])

# Process each class folder
for class_name in os.listdir(source_folder):
    class_path = os.path.join(source_folder, class_name)
    destination_class_path = os.path.join(destination_folder, class_name)

    if os.path.isdir(class_path):
        os.makedirs(destination_class_path, exist_ok=True)

        for filename in os.listdir(class_path):
            if filename.endswith(('.jpg', '.png', '.jpeg')):  # Adjust extensions if needed
                base_name, ext = filename.split('.')
                
                # Load image
                img_path = os.path.join(class_path, filename)
                image = cv2.imread(img_path)
                image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

                # Save original image first
                original_dest = os.path.join(destination_class_path, filename)
                cv2.imwrite(original_dest, cv2.cvtColor(image, cv2.COLOR_RGB2BGR))

                # Apply augmentation 3 times
                for i in range(3):  # Generate 3 augmented images
                    augmented = augmentation(image=image)["image"]
                    new_filename = f"{base_name}_{3 + i}.{ext}"  # X_3, X_4, X_5...
                    new_path = os.path.join(destination_class_path, new_filename)
                    cv2.imwrite(new_path, cv2.cvtColor(augmented, cv2.COLOR_RGB2BGR))

print("Augmented dataset created successfully in 'train_augmented'!")
```
