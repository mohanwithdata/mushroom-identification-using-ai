```python
import os
import shutil
import random

# Define paths
dataset_folder = r"D:\Project\data_kaggle\aug\train_augmented"  # Path to the dataset
train_folder = r"D:\Project\data_kaggle\aug\train_split"  # Output train folder
val_folder = r"D:\Project\data_kaggle\aug\val_split"  # Output validation folder

# Split ratio
train_ratio = 0.8  # 80% train, 20% validation

# Ensure train and validation folders exist
os.makedirs(train_folder, exist_ok=True)
os.makedirs(val_folder, exist_ok=True)

# Process each class folder
for class_name in os.listdir(dataset_folder):
    class_path = os.path.join(dataset_folder, class_name)
    train_class_path = os.path.join(train_folder, class_name)
    val_class_path = os.path.join(val_folder, class_name)

    if os.path.isdir(class_path):
        os.makedirs(train_class_path, exist_ok=True)
        os.makedirs(val_class_path, exist_ok=True)

        # Get all images in the class
        images = [img for img in os.listdir(class_path) if img.endswith(('.jpg', '.png', '.jpeg'))]
        
        # Shuffle images randomly
        random.shuffle(images)

        # Split into train and validation
        split_index = int(len(images) * train_ratio)
        train_images = images[:split_index]
        val_images = images[split_index:]

        # Copy images to respective folders
        for img in train_images:
            shutil.copy2(os.path.join(class_path, img), os.path.join(train_class_path, img))

        for img in val_images:
            shutil.copy2(os.path.join(class_path, img), os.path.join(val_class_path, img))

print("Dataset successfully split into 80% train and 20% validation!")
```
