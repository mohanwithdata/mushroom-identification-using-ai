```python
import os
import matplotlib.pyplot as plt
from PIL import Image
import random

# Set the path to your dataset
dataset_dir = r'D:\Project\data_kaggle\test'

# List all subdirectories (representing classes) in the dataset
class_dirs = os.listdir(dataset_dir)

# Set up the plot
fig, axes = plt.subplots(len(class_dirs), 5, figsize=(15, 3 * len(class_dirs)))

# Loop through each class directory and display 5 random images
for i, class_dir in enumerate(class_dirs):
    class_path = os.path.join(dataset_dir, class_dir)
    
    # Get all image files in the class directory
    image_files = [f for f in os.listdir(class_path) if f.endswith(('.jpg', '.jpeg', '.png'))]
    
    # Pick 5 random images to display
    sample_images = random.sample(image_files, 5)
    
    for j, image_file in enumerate(sample_images):
        image_path = os.path.join(class_path, image_file)
        img = Image.open(image_path)
        
        # Convert to RGB if the image is not in RGB mode
        if img.mode != 'RGB':
            img = img.convert('RGB')
        
        # Plot the image
        axes[i, j].imshow(img)
        axes[i, j].axis('off')  # Hide axis
        axes[i, j].set_title(f"{class_dir}", fontsize=10)

# Adjust the layout and save the figure as an image
plt.tight_layout()
plt.savefig('train_dataset_samples.png', bbox_inches='tight')  # Save the image to a file
plt.close()  # Close the plot to free up memory
```
