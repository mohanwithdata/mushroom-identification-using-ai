```python
import os
import shutil
import re
from collections import defaultdict

# Define source and destination folders
source_folder = r'C:\Users\mohan\Downloads\dataset_augmented_inception\test'  # Replace with your dataset path
destination_folder = r'D:\Project\data_kaggle\test'  # Folder to store filtered images

# Ensure the destination folder exists
os.makedirs(destination_folder, exist_ok=True)

# Process each class folder
for class_name in os.listdir(source_folder):
    class_path = os.path.join(source_folder, class_name)
    destination_class_path = os.path.join(destination_folder, class_name)

    if os.path.isdir(class_path):
        os.makedirs(destination_class_path, exist_ok=True)

        # Dictionary to store images grouped by base name (e.g., "0", "1", "2" etc.)
        image_groups = defaultdict(list)

        # Collect images
        for filename in sorted(os.listdir(class_path)):  # Sorting ensures consistency
            if filename.lower().endswith(('.jpg', '.png', '.jpeg')):  # Adjust extensions if needed
                base_index = filename.split('_')[0]  # Extract base index (e.g., "0" from "0_3.png")
                image_groups[base_index].append(filename)

        # Copy only original + 2 augmented images
        for base_index, images in image_groups.items():
            # Sort images numerically based on their second number (X_0, X_1, X_2, etc.)
            images = sorted(images, key=lambda x: int(x.split('_')[1].split('.')[0]))

            # Keep only first 3 images (original + 2 augmentations)
            selected_images = images[:1]

            for image in selected_images:
                src = os.path.join(class_path, image)
                dst = os.path.join(destination_class_path, image)
                shutil.copy2(src, dst)

print("Filtered dataset created successfully with original + 2 augmented images per base index in each class.")
```
